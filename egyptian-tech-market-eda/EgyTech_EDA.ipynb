```python
upper_limit = eda_df["yoe"].quantile(0.99)

eda_df["yoe_capped"] = np.where(
    eda_df["yoe"] > upper_limit,
    upper_limit,
    eda_df["yoe"]
)
```

```python
sns.boxplot(x="gender", y="yoe_capped", data=eda_df)
plt.title("YOE by Gender (Capped at 99th percentile)")
plt.show()
```

## Dataset: Egyptian Tech Market Survey 2024 (EgyTech FYI)

🔹 1. Demographic Insights

The dataset includes respondents from diverse educational backgrounds and job titles across the Egyptian tech industry.

Gender distribution indicates varied participation, enabling comparative analysis of experience and compensation.

Most professionals fall within early to mid-career stages, as shown by the Years of Experience (YOE) distribution.

🔹 2. Experience Analysis

Years of Experience shows a right-skewed distribution, with the majority of respondents having fewer than 10 years of experience.

Outliers represent senior professionals with extensive experience; these were retained as valid real-world observations.

Experience bucket analysis highlights a strong concentration in junior and mid-level roles.

🔹 3. Programming Language Trends

Python, SQL, and JavaScript emerge as the most commonly used programming languages.

Many respondents report using multiple programming languages, reflecting the multi-skill nature of modern tech roles.

Language usage trends align strongly with data, software, and web-focused roles.

🔹 4. Company & Market Insights

Small to medium-sized companies make up a large portion of the dataset, indicating a startup- and SME-driven tech ecosystem.

Respondents work across multiple business markets, though a significant portion did not specify market details.

Company size shows a clear relationship with experience and compensation levels.

🔹 5. Salary & Compensation Analysis (KEY HIGHLIGHT)

Salary distribution is right-skewed, with most professionals earning within mid-range compensation brackets.

Compensation increases steadily with higher Years of Experience, confirming expected career progression.

Larger organizations tend to offer higher and more consistent salary packages compared to smaller firms.

Salary bucket analysis provides a cleaner view of compensation trends across the industry.

🔹 6. Data Quality & Handling

The dataset contained missing values in select categorical fields (e.g., business market, programming languages).

Missing values were handled using context-aware strategies (replacement, selective exclusion for analysis).

No rows were dropped blindly, preserving the integrity of the survey data.
